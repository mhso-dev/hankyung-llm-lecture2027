# 도구 설치와 터미널

현재 작업 폴더와 상대 경로의 관계를 임시 폴더에서 확인합니다. 폴더는 파일과 다른 폴더를 담는 공간이고, 경로는 그 위치를 글자로 나타낸 값입니다. 상대 경로는 기준 폴더가 있어야 하나의 실제 위치를 가리킵니다.

## 실행 환경 준비

개인 파일을 바꾸지 않도록 Python이 제공하는 임시 폴더만 사용합니다. `from pathlib import Path`는 경로를 다루는 `Path`를, `from tempfile import TemporaryDirectory`는 임시 폴더를 만드는 기능을 가져옵니다. `TemporaryDirectory()`가 반환한 결과에는 임시 폴더 이름과 정리 기능이 들어 있습니다. `Path(temporary_folder.name)`은 그 이름을 경로 객체로 바꿉니다. `/`는 `Path` 뒤에 폴더 이름을 연결하고, `mkdir()`가 실제 폴더를 만듭니다.

```python
from pathlib import Path
from tempfile import TemporaryDirectory

# TemporaryDirectory는 실습이 끝나면 지워지는 임시 폴더를 만듭니다.
temporary_folder = TemporaryDirectory()
workspace = Path(temporary_folder.name)
project = workspace / "practice"
project.mkdir()
print(f"임시 실습 폴더 준비: {project.is_dir()}")
```

## 기본 실행: 상대 경로로 파일 준비

상대 경로 `data/message.txt`는 어느 폴더를 기준으로 하는지에 따라 다른 파일을 가리킵니다. 먼저 `project / "data"`로 경로를 만들고 `mkdir()`로 `data` 폴더를 만듭니다. 다시 `message.txt`를 연결한 뒤 `write_text`로 글을 저장합니다. `encoding="utf-8"`은 한글을 파일의 바이트로 바꾸고 다시 읽을 때 사용할 문자 규칙입니다. `is_file()`은 해당 위치가 파일이면 `True`를 반환합니다.

```python
# slash(/)는 기준 폴더 뒤에 폴더와 파일 이름을 연결합니다.
data_folder = project / "data"
data_folder.mkdir()
message_file = data_folder / "message.txt"
message_file.write_text("안녕, 터미널!", encoding="utf-8")
relative_path = Path("data/message.txt")
print(f"상대 경로: {relative_path}")
print(f"파일 준비: {message_file.is_file()}")
```

## 정상 경로 읽기

의도한 `project`를 기준으로 선택하면 파일을 읽을 수 있습니다. `project / relative_path`는 기준 폴더와 `data/message.txt`를 연결해 실제 파일 경로를 만듭니다. `read_text`는 파일 내용을 문자열로 반환하고, `basic_message`가 그 반환값을 저장합니다. 출력에서 저장한 문장과 읽은 문장이 같은지 확인합니다.

```python
# project와 상대 경로를 연결해 저장한 내용을 읽습니다.
basic_path = project / relative_path
basic_message = basic_path.read_text(encoding="utf-8")
print(f"기본 읽기: {basic_message}")
assert basic_message == "안녕, 터미널!"
```

## Before: 기준 폴더가 다른 경우

작업 기준을 `workspace`로 잘못 선택하면 같은 상대 경로에서 파일을 찾지 못합니다. `try` 아래에는 실패할 수 있는 읽기를 둡니다. `FileNotFoundError`가 발생하면 `except` 아래가 실행되고 오류의 종류를 출력합니다. `else`는 예상한 오류가 발생하지 않았을 때 실행됩니다. 이 구조는 Ch 2에서 자세히 배우며, 여기서는 경로 오류만 구분해 관찰하기 위해 사용합니다.

```python
# workspace 아래에는 data 폴더가 없으므로 예상된 오류가 발생합니다.
wrong_path = workspace / relative_path
try:
    wrong_path.read_text(encoding="utf-8")
except FileNotFoundError as error:
    print(f"실패 유형: {type(error).__name__}")
else:
    raise AssertionError("잘못된 기준에서는 파일이 없어야 합니다.")
```

## After: 프로젝트 폴더를 기준으로 선택

같은 상대 경로를 `project`에 연결하면 저장한 파일을 읽을 수 있습니다. 파일 이름과 내용은 바꾸지 않고 기준 폴더만 `workspace`에서 `project`로 수정합니다. `cleanup()`은 실습용 임시 폴더를 정리합니다. 출력이 `안녕, 터미널!`인지 확인한 뒤 정리하므로, 읽기 성공과 정리를 한 단계씩 구분할 수 있습니다.

```python
# project를 기준으로 같은 상대 경로를 연결합니다.
right_path = project / relative_path
message = right_path.read_text(encoding="utf-8")
print(f"읽은 내용: {message}")
assert message == "안녕, 터미널!"
temporary_folder.cleanup()
```
