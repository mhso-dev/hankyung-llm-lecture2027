# 과정 지도와 학습 규칙

예상하고, 실행하고, 값을 바꾼 뒤 다시 확인하는 학습 순서를 연습합니다. 프로그램은 컴퓨터가 수행할 작업을 적은 것이고, 코드는 그 작업을 Python 문법으로 표현한 글입니다. 코드 셀을 실행하면 커널이 코드를 읽고 앞 셀에서 만든 값을 기억합니다.

## 실행 환경 확인

이 노트북은 Python 3.12 커널에서 첫 셀부터 순서대로 실행합니다. 커널은 노트북의 Python 코드를 실행하고 앞 셀에서 만든 값을 기억하는 프로그램입니다. `import sys`는 Python에 포함된 `sys` 기능을 가져오며, `sys.version_info[:2]`는 실행 중인 Python 버전에서 앞의 두 숫자를 고릅니다. `== (3, 12)`의 결과가 `True`이면 주 버전 3과 부 버전 12가 맞습니다. `assert`는 이 조건이 거짓일 때 실행을 멈춥니다.

```python
import sys

# version_info는 현재 Python의 주 버전과 부 버전을 숫자로 제공합니다.
python_312 = sys.version_info[:2] == (3, 12)
print(f"Python 3.12: {python_312}")
assert python_312
```

## 1단계: 결과 예상

가격이 3,000원인 상품 2개의 총액을 먼저 예상합니다. `price`, `count`, `prediction`은 각각 단가, 수량, 예상 합계에 붙인 이름입니다. `=`는 오른쪽 값을 왼쪽 이름에 저장합니다. 문자열 앞의 `f`와 문장 안의 중괄호는 저장한 값을 출력 문장에 넣습니다. 이 문법은 Ch 2에서 배우며, 여기서는 세 이름에 어떤 값이 저장되는지 한 줄씩 확인합니다.

```python
# price는 한 개의 가격, count는 상품 개수입니다.
price = 3000
count = 2
prediction = 6000
print(f"예상 총액: {prediction}원")
```

## 2단계: 코드 실행

곱셈 기호 `*`로 실제 총액을 계산하고 예상과 맞는지 확인합니다. `total = price * count`는 저장된 3,000과 2를 곱해 6,000을 만든 뒤 `total`에 저장합니다. `print`는 값을 화면에 보여 주고, `assert total == prediction`은 실제 합계와 예상 합계가 같은지 검사합니다. 조건이 참이면 `assert`는 아무 문장도 출력하지 않습니다.

```python
# price * count의 결과를 total이라는 이름으로 저장합니다.
total = price * count
print(f"실행 총액: {total}원")
assert total == prediction
```

## 3단계: 값 하나 변경

가격은 그대로 두고 개수만 3개로 바꿥니다. 결과가 9,000원으로 변하는지 비교합니다.

```python
# count만 2에서 3으로 바꾼 뒤 다시 계산합니다.
count = 3
changed_total = price * count
print(f"2개: {total}원")
print(f"3개: {changed_total}원")
assert changed_total == 9000
```

## 실험 결과 해석

예상과 실행 결과가 6,000원으로 같았습니다. 개수 하나만 바꾸었을 때 총액은 9,000원으로 변했습니다. `assert`는 관찰한 값이 기대한 값과 같은지 확인합니다.
