# 변수와 자료형

이름, 수량, 가격처럼 성격이 다른 값을 변수에 저장하고 계산 결과를 확인합니다.

## 실행 환경 준비

이 노트북은 Python 3.12 표준 기능만 사용합니다. 첫 셀부터 순서대로 실행합니다. `이름 = 값`은 오른쪽 값을 왼쪽 변수에 저장한다는 뜻입니다. 따옴표로 감싼 값은 문자열이고, 따옴표 없는 숫자는 계산에 사용할 수 있는 수입니다.

```python
# 실습에서 사용할 상품 정보를 서로 다른 자료형으로 저장합니다.
product_name = "무선 키보드"
quantity = 2
unit_price = 39000.0
is_member = True
print(product_name, quantity, unit_price, is_member)
```

## 기본 실행

수량과 단가를 곱해 주문 금액을 계산합니다. `*`는 곱셈 기호입니다. `print(...)`처럼 이름 뒤에 소괄호를 붙이면 함수를 호출합니다. `type(quantity)`는 `quantity`에 저장된 값의 자료형을 반환하고, 뒤의 `.__name__`은 그 자료형의 이름만 읽습니다. `f"...{변수}..."`는 중괄호 자리에 변수 값을 넣어 출력할 문자열을 만드는 표기입니다.

```python
# 변수 이름으로 값을 다시 사용해 주문 금액을 계산합니다.
order_total = quantity * unit_price
print(f"상품: {product_name}")
print(f"주문 금액: {order_total:.0f}원")
print(type(quantity).__name__, type(unit_price).__name__)
```

## 중간 결과 관찰

회원 할인율을 10%로 저장하고, 계산에 사용한 값과 결과를 함께 확인합니다. 입력은 주문 금액 `order_total`과 할인율 `discount_rate`입니다. 두 값을 곱해 할인액을 처리하고, 주문 금액에서 할인액을 빼 결제 금액을 만듭니다. `:.0%`는 0.1을 10%로, `:.0f`는 소수점 아래 없이 표시하라는 출력 형식입니다. 저장된 값 자체를 바꾸지는 않습니다.

```python
# 할인율을 소수 0.1로 저장해 10% 할인을 계산합니다.
discount_rate = 0.1
discount = order_total * discount_rate
payment = order_total - discount
print(f"회원 여부: {is_member}")
print(f"할인율: {discount_rate:.0%}")
print(f"결제 금액: {payment:.0f}원")
```

## 값 하나를 바꿔 비교

상품과 단가는 그대로 두고 수량만 3으로 바꿉니다. 원래 변수 `quantity`를 덮어쓰지 않고 `changed_quantity`라는 새 변수에 저장하므로 변경 전후를 같은 출력에서 비교할 수 있습니다.

```python
# 수량만 바꿔 기존 결과와 새 결과를 나란히 출력합니다.
changed_quantity = 3
changed_total = changed_quantity * unit_price
print(f"수량 {quantity}: {order_total:.0f}원")
print(f"수량 {changed_quantity}: {changed_total:.0f}원")
```

## 실패 재현

입력으로 받은 수량은 문자열입니다. `input_quantity = "2"`의 따옴표가 이 값을 숫자 2가 아닌 글자 `2`로 만듭니다. 문자열에 정수를 곱하는 처리는 문자열 반복이므로 글자가 세 번 이어진 `"222"`를 반환합니다. 오류 메시지는 없지만 주문 수량 계산으로는 틀린 결과입니다.

```python
# 문자열 곱셈이 만든 잘못된 주문 수량을 관찰합니다.
input_quantity = "2"
wrong_quantity = input_quantity * 3
print(f"자료형: {type(input_quantity).__name__}")
print(f"계산 결과: {wrong_quantity}")
```

## 수정

`int()`는 숫자 모양의 문자열을 입력받아 정수를 반환합니다. 반환된 `2`를 `converted_quantity`에 저장한 뒤 3을 곱하면 정수 `6`이 됩니다. 마지막 `assert correct_quantity == 6`은 `==`의 양쪽 값이 같은지 검사하며, 조건이 거짓이면 실행을 실패시킵니다.

```python
# 숫자 모양의 문자열을 정수로 변환한 뒤 계산합니다.
input_quantity = "2"
converted_quantity = int(input_quantity)
correct_quantity = converted_quantity * 3
print(f"변환 전: {type(input_quantity).__name__}")
print(f"변환 후: {type(converted_quantity).__name__}")
print(f"계산 결과: {correct_quantity}")
assert correct_quantity == 6
```
