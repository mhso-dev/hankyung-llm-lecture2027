<h3>개념 요약</h3>
<p>임베딩은 문장을 숫자 벡터로 바꾸고, 벡터스토어는 질문 벡터와 가까운 문서를 찾습니다. 세 문서만 색인해 의미가 가까운 결과가 먼저 나오는지 확인합니다.</p>

```python
# os로 API 키 환경 변수를 읽고, find_dotenv(usecwd=True)로 프로젝트 루트의 .env를 찾습니다.
# env_file은 찾은 파일 경로이며 load_dotenv()는 그 파일의 값을 현재 프로세스에 로드합니다.
# assert는 .env와 OPENAI_API_KEY 중 하나라도 없으면 실제 호출 전에 실행을 멈춥니다.
import os

from dotenv import find_dotenv, load_dotenv

env_file = find_dotenv(usecwd=True)
assert env_file and load_dotenv(env_file), '프로젝트 루트 .env를 찾지 못했습니다.'
assert os.getenv('OPENAI_API_KEY'), 'OPENAI_API_KEY가 없습니다.'
print('환경 준비 완료')
```

<h3>예제 실행</h3>
<p>서로 다른 주제의 짧은 문서 세 건과 문서 ID를 준비합니다.</p>

```python
# texts의 각 튜플은 (문서 ID, 본문) 두 값을 묶어 보관합니다.
# 리스트 표기식은 for로 texts를 순회하며 각 튜플을 i와 t로 구조 분해합니다.
# docs에는 본문과 doc_id 메타데이터를 가진 Document 세 건이 남습니다.
from langchain_core.documents import Document
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_openai import OpenAIEmbeddings

texts = [('D01', '파란도서관의 정기 휴관일은 매주 월요일입니다.'),
         ('D02', '해솔식당의 점심 영업은 오전 11시에 시작합니다.'),
         ('D03', '별빛회사의 급여일은 매월 25일입니다.')]
docs = [Document(page_content=t, metadata={'doc_id': i}) for i, t in texts]
```

<h3>실제 임베딩 검색</h3>
<p>OpenAI 임베딩으로 문서를 색인하고 도서관 질문과 가까운 결과를 출력합니다.</p>

```python
# OpenAIEmbeddings의 model 인자는 문장을 벡터로 바꿀 임베딩 모델을 선택합니다.
# store.add_documents()는 docs를 임베딩해 메모리 벡터스토어에 추가합니다.
# similarity_search_with_score()는 상위 k=2의 (Document, 유사도 점수) 튜플을 반환합니다.
embedding = OpenAIEmbeddings(model='text-embedding-3-small')
store = InMemoryVectorStore(embedding=embedding)
store.add_documents(docs)
hits = store.similarity_search_with_score('도서관이 쉬는 요일은?', k=2)
for doc, score in hits:
    print(doc.metadata['doc_id'], round(score, 3), doc.page_content)
```

<h3>변형 실행</h3>
<p>질문을 급여일 질문으로 바꾼 뒤 첫 번째 문서 ID를 확인합니다.</p>

```python
# similarity_search()는 점수 없이 Document만 반환하며 k=1은 첫 결과 한 건을 요청합니다.
# pay_hits[0]은 첫 문서이고, metadata['doc_id']는 그 문서의 식별자를 읽습니다.
pay_hits = store.similarity_search('월급은 며칠에 받나요?', k=1)
print(pay_hits[0].metadata['doc_id'])
```

<h3>실행 검증</h3>
<p>앞 셀의 결과가 이 절에서 확인할 조건을 만족하는지 단언문으로 검사합니다.</p>

```python
# 점수가 포함된 hits에서 [0][0]은 첫 튜플의 Document를 선택합니다.
# 두 assert는 도서관 질문과 급여일 질문의 첫 문서 ID를 각각 확인합니다.
assert hits[0][0].metadata['doc_id'] == 'D01'
assert pay_hits[0].metadata['doc_id'] == 'D03'
print('검증 통과: 두 질문이 알맞은 문서를 검색했습니다.')
```

<h3>정리</h3>
<p>임베딩 모델은 단어가 정확히 같지 않아도 의미가 가까운 문서를 찾습니다. 검색 점수는 모델과 벡터스토어에 따라 해석이 달라지므로 같은 설정 안에서 순위를 비교합니다.</p>

<p><strong>원인</strong> — <code>k</code>에 정수 대신 문자열을 전달하면 검색 결과를 자르는 단계에서 타입 오류가 발생합니다.</p>

```python
# k는 결과 개수이므로 정수가 필요하지만 의도적으로 'two'를 전달합니다.
# try/except는 발생한 TypeError만 포착해 Run All을 중단하지 않고 오류 메시지를 보여 줍니다.
try:
    store.similarity_search('휴관일', k='two')
except TypeError as error:
    print('예상한 실패:', error)
```

<p><strong>수정</strong> — 검색 개수를 정수 <code>2</code>로 전달하여 같은 질문을 실행합니다.</p>

```python
# fixed_hits는 k를 정수 2로 수정해 받은 Document 리스트입니다.
# 리스트 표기식으로 각 문서의 doc_id를 fixed_ids에 모읍니다.
# 결과가 두 건이고 휴관일 문서 D01이 포함되었는지 검사합니다.
fixed_hits = store.similarity_search('휴관일', k=2)
fixed_ids = [doc.metadata['doc_id'] for doc in fixed_hits]
print('After:', fixed_ids)
assert len(fixed_hits) == 2
assert 'D01' in fixed_ids
```
