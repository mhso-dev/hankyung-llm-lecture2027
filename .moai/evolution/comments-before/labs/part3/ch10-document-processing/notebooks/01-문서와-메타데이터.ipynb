<h3>문서와 메타데이터</h3>
<p><strong>개념</strong> 문서 본문과 출처 정보는 함께 이동해야 합니다. 서로 다른 로더의 필드 이름을 공통 계약으로 바꾸면 이후 청킹과 검색 결과를 같은 방식으로 추적할 수 있습니다.</p>

```python
import sys
print(sys.version.split()[0])
```

```python
from pathlib import Path
source = Path("규정.pdf")
```

<h3>공통 메타데이터 만들기</h3>
<p><strong>확인할 점</strong> <code>source</code>에는 컴퓨터의 절대 경로 대신 공유 가능한 파일명이나 프로젝트 상대 경로를 저장합니다.</p>

```python
raw = {"page": "2", "tags": ["배송", "환불"]}
metadata = {
    "doc_id": source.stem, "source": source.name,
    "file_type": source.suffix.lstrip(".").lower(),
    "parser": "notebook-demo", "page": int(raw["page"]),
    "tags": ",".join(raw["tags"]),
}
metadata
```

<h3>일부러 실패해 보기</h3>
<p><strong>실패 사례</strong> 쪽 번호가 0이면 출처 위치를 잘못 가리킵니다. 다음 셀은 오류를 포착해 메시지로 확인하므로 노트북 실행이 중단되지 않습니다.</p>

```python
try:
    page = 0
    if page < 1:
        raise ValueError("page는 1 이상이어야 합니다")
except ValueError as error:
    print(f"caught: {error}")
```

<p><strong>수정 이유</strong> 쪽 번호를 1부터 시작하는 값으로 검증한 뒤 메타데이터에 저장합니다.</p>

```python
page = 1
assert page >= 1
fixed_metadata = {**metadata, "page": page}
print("validated page:", fixed_metadata["page"])
```
