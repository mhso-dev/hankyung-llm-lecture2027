# Command로 상태 갱신과 이동

한 노드가 판정 결과를 상태에 기록하면서 다음 노드까지 함께 정할 수 있을까요?

## 개념을 코드에 연결하기

**호출 표기와 실행 시점.** Command(update={...}, goto='generate')는 상태 갱신과 다음 목적지를 하나의 반환값에 담습니다. 노드가 이 객체를 반환하면 런타임은 update를 상태에 반영한 뒤 goto가 가리키는 노드로 이동합니다. 반환형 Command[Literal['generate', 'request_more']]는 가능한 목적지를 실행 전에 읽을 수 있게 합니다.

**다른 상황에 적용.** 승인 흐름에서 reviewer 노드가 {'decision': 'approved'}를 남기면서 publish로 이동하거나, 반려 사유를 남기면서 revise로 이동할 수 있습니다. 판단 기록과 경로가 같은 결정에서 만들어져야 할 때 Command가 잘 맞습니다.

**비교와 경계.** 조건부 엣지는 경로 규칙을 그래프 조립부에 두고 Command는 노드 반환값 안에 둡니다. 경로를 한곳에서 살펴보는 것이 중요하면 조건부 엣지가 낫고, 갱신과 이동이 분리되면 불일치하기 쉬운 경우에는 Command가 낫습니다.

## 상태와 Command 노드 정의

다음 코드에서 상태와 Command 노드 정의에 필요한 형식과 함수를 선언합니다. 이 셀은 정의를 준비하며, 값은 이어지는 실행 셀에서 확인합니다.

**이 셀의 코드 읽기.** 이 셀을 실행하면 `DecisionState` 클래스 정의가 준비됩니다. 클래스 본문은 정의 시 처리되지만 인스턴스 값은 나중에 객체를 만들 때 생깁니다. `route(state: DecisionState)`는 함수를 정의하는 표기입니다. 이 셀에서는 함수 객체만 준비되고 들여쓴 본문은 뒤에서 함수를 호출할 때 실행됩니다. `return` 뒤의 값은 화면 출력이 아니라 호출한 코드로 전달되는 반환값입니다. `Command(update=..., goto=...)`는 상태 갱신값과 다음 목적지를 함께 반환합니다.


```python
# 상태와 Command 노드 정의에 사용할 상태 형식과 함수를 선언합니다.
from typing import Literal
from typing_extensions import TypedDict
from langgraph.graph import StateGraph, START, END
from langgraph.types import Command

class DecisionState(TypedDict):
    score: float
    decision: str
    answer: str

def route(state: DecisionState) -> Command[Literal['generate', 'request_more']]:
    enough = state['score'] >= 0.7
    return Command(
        update={'decision': 'sufficient' if enough else 'insufficient'},
        goto='generate' if enough else 'request_more',
    )
```

## 목적지 노드와 그래프 정의

다음 코드에서 목적지 노드와 그래프 정의에 필요한 형식과 함수를 선언합니다. 이 셀은 정의를 준비하며, 값은 이어지는 실행 셀에서 확인합니다.

**이 셀의 코드 읽기.** `generate(state: DecisionState)`, `request_more(state: DecisionState)`는 함수를 정의하는 표기입니다. 이 셀에서는 함수 객체만 준비되고 들여쓴 본문은 뒤에서 함수를 호출할 때 실행됩니다. `return` 뒤의 값은 화면 출력이 아니라 호출한 코드로 전달되는 반환값입니다. `StateGraph(StateType)`은 상태 형식을 입력으로 받아 그래프 설계 객체를 만듭니다. `add_node(name, function)`은 그래프 이름과 실제 실행할 함수를 연결하며, 이 호출만으로 함수 본문이 실행되지는 않습니다. `add_edge(from, to)`는 두 노드의 고정 이동 순서를 등록합니다.


```python
# 목적지 노드와 그래프 정의에 사용할 상태 형식과 함수를 선언합니다.
def generate(state: DecisionState) -> dict:
    return {'answer': '선택한 근거로 답변을 생성합니다.'}

def request_more(state: DecisionState) -> dict:
    return {'answer': '답변 전에 자료를 추가로 확인합니다.'}

builder = StateGraph(DecisionState)
builder.add_node('route', route)
builder.add_node('generate', generate)
builder.add_node('request_more', request_more)
builder.add_edge(START, 'route')
builder.add_edge('generate', END)
builder.add_edge('request_more', END)
graph = builder.compile()
```

## 두 경로 실행

두 경로 실행 단계를 실행합니다. 출력된 상태 값과 코드에서 반환한 키를 서로 대조합니다.

**이 셀의 코드 읽기.** `invoke(input)`가 나타나는 줄에서 실제 그래프 또는 모델 실행이 시작되고 반환값이 왼쪽 변수에 저장됩니다. `print()`는 확인용 문자열을 화면에 표시할 뿐 함수의 반환값을 바꾸지 않습니다. `assert`는 조건이 참이면 출력 없이 통과하고 거짓이면 실행을 중단하므로, 바로 앞 결과를 검증하는 역할을 합니다.


```python
# 두 경로 실행 단계의 연산을 실행하고 반환된 상태 값을 확인합니다.
base = {'decision': '', 'answer': ''}
high = graph.invoke({**base, 'score': 0.82})
low = graph.invoke({**base, 'score': 0.41})
print('0.82:', high['decision'], high['answer'])
print('0.41:', low['decision'], low['answer'])
assert high['decision'] == 'sufficient'
assert low['decision'] == 'insufficient'
```

## 가능한 이동 시각화

컴파일한 그래프에서 Mermaid 문자열을 생성합니다. 출력에서 노드 이름과 연결 방향을 확인합니다.

**이 셀의 코드 읽기.** `draw_mermaid()`는 컴파일된 그래프의 노드와 연결을 Mermaid 문자열로 반환하며 그래프를 다시 실행하지 않습니다. `print()`는 확인용 문자열을 화면에 표시할 뿐 함수의 반환값을 바꾸지 않습니다. `assert`는 조건이 참이면 출력 없이 통과하고 거짓이면 실행을 중단하므로, 바로 앞 결과를 검증하는 역할을 합니다.


```python
# 컴파일된 그래프의 실제 노드와 엣지를 Mermaid 문자열로 확인합니다.
mermaid = graph.get_graph().draw_mermaid()
print(mermaid)
assert 'generate' in mermaid and 'request_more' in mermaid
```
