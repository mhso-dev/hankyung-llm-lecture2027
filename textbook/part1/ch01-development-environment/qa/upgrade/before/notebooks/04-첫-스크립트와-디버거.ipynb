# 첫 스크립트와 디버거

같은 가격과 개수로 잘못된 덧셈 결과를 디버거에서 확인하고 곱셈으로 수정합니다. 스크립트는 Python 명령을 순서대로 적어 저장한 .py 파일이고, 디버거는 실행을 특정 줄에서 멈춰 그 시점의 값을 확인하는 도구입니다.

## 실행 환경 준비

개인 파일을 바꾸지 않도록 임시 폴더에 Python 스크립트를 작성합니다. `import`는 준비된 기능을 현재 코드로 가져옵니다. `subprocess`는 다른 프로그램을 실행하고, `sys`는 현재 Python의 정보를 제공하며, `Path`는 경로를 다룹니다. `bug_file`은 `price_bug.py`가 놓일 경로만 나타내며, 이 셀에서는 아직 파일을 만들지 않습니다.

```python
import subprocess
import sys
from pathlib import Path
from tempfile import TemporaryDirectory

# 임시 폴더는 노트북 실행이 끝나면 삭제됩니다.
temporary_folder = TemporaryDirectory()
workspace = Path(temporary_folder.name)
bug_file = workspace / "price_bug.py"
```

## 기본 실행: 정상 총액 계산

가격과 개수를 곱해 6,000원이 나오는 기본 코드를 먼저 확인합니다. `=`는 오른쪽 값을 왼쪽 이름에 저장하고, `*`는 곱셈을 뜻합니다. `print`는 계산 결과를 화면에 표시합니다. `assert`는 `total`이 6000인지 검사하며, 조건이 참이면 별도 문장을 출력하지 않습니다.

`+`나 `*`처럼 계산 방법을 지정하는 기호를 연산자라고 합니다. 이 실습은 더하기 기호를 곱하기 기호로 바꾸었을 때 총액이 어떻게 달라지는지 확인합니다.

```python
# price는 한 개의 가격, count는 상품 개수입니다.
price = 3000
count = 2
total = price * count
print(f"정상 총액: {total}원")
assert total == 6000
```

## Before: 덧셈으로 작성한 스크립트

가격 3,000원과 개수 2개를 `+`로 계산하면 코드는 실행되지만 총액은 틀립니다. 삼중 큰따옴표는 여러 줄의 Python 코드를 하나의 문자열로 담습니다. `write_text`는 그 문자열을 파일에 저장합니다. `subprocess.run`은 현재 노트북의 Python으로 저장한 파일을 실행하고 결과 객체를 반환합니다. `stdout`은 스크립트가 출력한 글이며, `strip`은 끝의 줄바꿈을 제거하고 `int`는 숫자 모양의 글을 정수로 바꿉니다.

```python
# write_text는 문자열을 Python 파일로 저장하고 쓴 문자 수를 반환합니다.
bug_source = """price = 3000
count = 2
total = price + count
print(total)
"""
bug_file.write_text(bug_source, encoding="utf-8")
bug_run = subprocess.run([sys.executable, bug_file], capture_output=True, text=True, check=True)
bug_total = int(bug_run.stdout.strip())
print(f"잘못된 총액: {bug_total}원")
assert bug_total == 3002
```

## 디버거로 중간 값 확인

`pdb`에서 4번 줄 직전에 멈춘 뒤 `price`, `count`, `total`을 직접 출력합니다. `b 4`는 4번 줄에 중단점을 두고, `c`는 그 줄까지 계속 실행하며, `p` 뒤의 이름은 현재 값을 출력합니다. `q`는 디버거를 종료합니다. 이 명령 문자열은 `input` 인자로 `pdb`에 전달됩니다. 출력 정리 코드는 경로와 안내 문장을 제외하고 마지막 숫자 세 개만 고르며, 목록 문법은 Ch 2에서 배웁니다.

```python
# pdb에 중단점, 계속 실행, 변수 출력, 종료 명령을 순서대로 전달합니다.
debug_commands = """b 4
c
p price
p count
p total
q
"""
debug_run = subprocess.run(
    [sys.executable, "-m", "pdb", bug_file.name],
    cwd=workspace, input=debug_commands, capture_output=True, text=True, check=True,
)
# pdb 프롬프트 뒤의 숫자 결과만 선택하여 개인 경로를 제외합니다.
observed = [line.removeprefix("(Pdb) ") for line in debug_run.stdout.splitlines()
            if line.removeprefix("(Pdb) ").isdigit()]
print(f"price = {observed[-3]}")
print(f"count = {observed[-2]}")
print(f"total = {observed[-1]}")
assert observed[-3:] == ["3000", "2", "3002"]
```

## After: 곱셈으로 수정

값은 그대로 두고 연산자만 `+`에서 `*`로 바꿥니다.

```python
# 가격과 개수를 곱하도록 연산자 하나만 수정합니다.
fixed_file = workspace / "hello.py"
fixed_source = bug_source.replace("price + count", "price * count")
fixed_file.write_text(fixed_source, encoding="utf-8")
fixed_run = subprocess.run([sys.executable, fixed_file], capture_output=True, text=True, check=True)
fixed_total = int(fixed_run.stdout.strip())
print(f"수정한 총액: {fixed_total}원")
assert fixed_total == 6000
temporary_folder.cleanup()
```

## 결과 해석

디버거에서 입력값 3,000과 2가 예상대로였지만 `total`이 3,002였음을 확인했습니다. 연산자를 곱셈으로 바꾼 뒤 같은 입력에서 6,000원을 확인했습니다. 오류 메시지가 없다는 사실은 문법상 실행되었다는 뜻이고, 요구한 총액이 맞는지는 예상값과 별도로 비교해야 합니다.
