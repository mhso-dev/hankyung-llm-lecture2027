# 자료구조

목록, 튜플, 딕셔너리, 집합에 같은 주문 정보를 서로 다른 목적에 맞춰 저장합니다.

## 실행 환경 준비

이 노트북은 Python 3.12 표준 기능만 사용합니다. 네 자료구조의 출력 모양과 값을 찾는 방법을 비교합니다. 괄호 모양만 외우지 말고 순서가 필요한지, 값을 바꿀지, 이름으로 찾을지, 중복을 없앨지를 먼저 판단합니다.

```python
# 같은 주문을 목적에 맞는 네 자료구조로 표현합니다.
items = ["키보드", "마우스", "키보드"]
delivery_point = (37.5665, 126.9780)
order = {"id": "A102", "amount": 78000}
unique_items = set(items)
print(items)
print(delivery_point)
print(order)
print(sorted(unique_items))
```

## 기본 실행

각 자료구조에서 값을 읽는 방법을 확인합니다. `items[0]`의 대괄호는 목록의 위치 0을 읽고, `order['amount']`는 딕셔너리에서 `amount` 키와 짝인 값을 읽습니다. `'마우스' in unique_items`는 집합에 해당 값이 있는지 검사해 불리언을 반환합니다. 집합은 순서를 보장하지 않으므로 `sorted()`에 넣어 정렬된 새 목록을 반환받은 뒤 출력합니다.

```python
# 위치, 키, 포함 여부에 맞는 읽기 방법을 사용합니다.
print(f"첫 상품: {items[0]}")
print(f"위도: {delivery_point[0]}")
print(f"주문 금액: {order['amount']}원")
print(f"마우스 포함: {'마우스' in unique_items}")
```

## 중간 결과 관찰

목록에는 중복 상품이 남고 집합에는 고유한 상품만 남습니다. `len()`은 묶음을 입력받아 항목 수를 정수로 반환합니다. `order['item_count'] = item_count`는 딕셔너리에 새 키와 값을 추가합니다. 목록의 길이 3은 담긴 상품 수이고 집합의 길이 2는 서로 다른 상품 종류 수이므로 서로 다른 질문의 답입니다.

```python
# 길이를 비교해 중복 제거 결과를 확인합니다.
item_count = len(items)
unique_count = len(unique_items)
order['item_count'] = item_count
print(f"담긴 상품 수: {item_count}")
print(f"서로 다른 상품 수: {unique_count}")
print(order)
```

## 값 하나를 바꿔 비교

`items.copy()`가 반환한 새 목록을 `changed_items`에 저장한 뒤 `append('헤드셋')`으로 끝에 값을 하나 추가합니다. 원본 `items`는 유지됩니다. 변경한 목록을 `set()`에 입력하면 중복을 제거한 새 집합이 반환됩니다.

```python
# 원본을 보존하려고 복사본에 값 하나를 추가합니다.
changed_items = items.copy()
changed_items.append("헤드셋")
changed_unique_items = set(changed_items)
print(f"변경 전: {items}")
print(f"변경 후: {changed_items}")
print(f"고유 상품: {sorted(changed_unique_items)}")
```

## 실패 재현

객체는 실행 중 만들어진 값의 실체이고 변수 이름은 그 객체를 가리킵니다. 같은 내용을 가진 별개 객체와, 한 객체를 가리키는 여러 이름을 구분합니다.

`backup = cart`는 `cart`의 내용을 복사하는 문장이 아닙니다. 두 변수 이름이 같은 목록 객체를 가리키게 합니다. 따라서 `cart.append('헤드셋')`으로 객체 하나를 수정하면 `backup`을 통해 읽은 내용에도 헤드셋이 나타납니다. `cart is backup`은 두 변수가 같은 객체를 가리키는지 검사합니다.

```python
# backup과 cart가 같은 목록 객체를 가리키는 상태를 만듭니다.
cart = ["키보드", "마우스"]
backup = cart
cart.append("헤드셋")
print(f"cart: {cart}")
print(f"backup: {backup}")
print(f"같은 객체: {cart is backup}")
```

## 수정

`copy()`는 현재 항목을 담은 새 목록을 반환합니다. `backup = cart.copy()`로 반환값을 저장하면 두 변수는 서로 다른 목록 객체를 가리킵니다. 이후 원본만 수정하므로 백업은 수정 전 값 두 개를 유지하고, `cart is backup`은 `False`를 반환합니다.

```python
# copy로 서로 다른 목록 객체를 만듭니다.
cart = ["키보드", "마우스"]
backup = cart.copy()
cart.append("헤드셋")
print(f"cart: {cart}")
print(f"backup: {backup}")
print(f"같은 객체: {cart is backup}")
assert backup == ["키보드", "마우스"]
```
