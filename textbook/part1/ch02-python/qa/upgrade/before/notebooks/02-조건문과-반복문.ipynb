# 조건문과 반복문

조건에 따라 배송비를 정하고, 여러 주문 금액을 순서대로 처리합니다.

## 실행 환경 준비

이 노트북은 Python 3.12 표준 기능만 사용합니다. 각 셀은 앞 셀의 결과를 이어서 사용합니다. 대괄호 안의 값은 다음 절에서 자세히 배우는 목록입니다. 여기서는 여러 금액을 앞에서부터 하나씩 읽기 위한 묶음으로 사용합니다.

```python
# 세 건의 주문 금액과 무료 배송 기준을 준비합니다.
order_amounts = [18000, 42000, 65000]
free_shipping_minimum = 50000
print(order_amounts)
print(f"무료 배송 기준: {free_shipping_minimum}원")
```

## 기본 실행

주문 금액 다섯 개를 `if`, `elif`, `else`의 세 경로와 차례로 비교합니다. `for amount in test_amounts:`는 목록에서 값을 하나 꺼내 `amount`에 저장한 뒤 들여쓴 블록을 반복한다는 뜻입니다. 콜론 `:`은 다음 들여쓰기 블록이 시작됨을 표시합니다. `>=`는 왼쪽 값이 오른쪽 값보다 크거나 같은지 검사해 `True` 또는 `False`를 만듭니다. 위에서 처음 참이 된 블록 하나만 실행합니다.

```python
# 세 경로와 50000원 경계를 실제 값으로 확인합니다.
test_amounts = [18000, 42000, 49999, 50000, 65000]
for amount in test_amounts:
    if amount >= free_shipping_minimum:
        shipping_fee = 0
    elif amount >= 30000:
        shipping_fee = 2000
    else:
        shipping_fee = 3000
    print(f"{amount}원 -> 배송비 {shipping_fee}원")
```

## 중간 결과 관찰

`while`은 조건이 참인 동안 반복합니다. `len(amounts_with_stop)`은 목록의 항목 수 5를 반환하고, `amounts_with_stop[index]`는 위치 번호에 해당하는 값을 읽습니다. 위치 번호는 0부터 시작합니다. `index += 1`은 `index = index + 1`과 같은 뜻입니다. 먼저 위치를 늘린 뒤 음수는 `continue`로 현재 차례의 나머지를 건너뛰고, 0은 `break`로 반복 전체를 끝냅니다. `append()`는 처리한 금액을 빈 목록의 끝에 추가합니다.

```python
# 위치 번호가 목록 길이보다 작은 동안 주문을 확인합니다.
amounts_with_stop = [18000, -1, 42000, 0, 65000]
index = 0
processed = []
while index < len(amounts_with_stop):
    amount = amounts_with_stop[index]
    index += 1
    if amount < 0:
        continue
    if amount == 0:
        break
    processed.append(amount)
print(f"처리한 주문: {processed}")
```

## 값 하나를 바꿔 비교

주문 금액은 그대로 두고 무료 배송 기준만 40000원으로 낮춥니다. 두 개의 개수 변수를 0에서 시작하고, 조건이 참일 때마다 `+= 1`로 한 건씩 더합니다. 이 셀은 값을 반환하는 함수가 없으며, 반복이 끝난 뒤 두 변수에 누적된 결과를 출력합니다.

```python
# 기준값 하나만 바꿔 결과를 비교합니다.
changed_minimum = 40000
free_count_before = 0
free_count_after = 0
for amount in order_amounts:
    if amount >= free_shipping_minimum:
        free_count_before += 1
    if amount >= changed_minimum:
        free_count_after += 1
print(f"기준 50000원: {free_count_before}건")
print(f"기준 40000원: {free_count_after}건")
```

## 실패 재현

목록 길이가 3이면 사용할 수 있는 위치 번호는 0, 1, 2입니다. `len(order_amounts) + 1`은 4이고, `range(4)`는 0, 1, 2, 3을 차례로 제공합니다. 마지막 `order_amounts[3]`은 존재하지 않는 네 번째 항목을 요청하므로 `IndexError`가 발생합니다. `try`는 실행을 시도할 블록, `except IndexError as error`는 해당 오류가 생겼을 때만 실행할 블록입니다. 예외 처리는 05절에서 자세히 배웁니다.

```python
# 목록 길이보다 한 칸 더 접근해 실제 IndexError를 관찰합니다.
try:
    for index in range(len(order_amounts) + 1):
        print(index, order_amounts[index])
except IndexError as error:
    print(type(error).__name__)
    print(error)
```

## 수정

`enumerate(order_amounts)`는 목록을 입력받아 실제 항목마다 `(위치 번호, 값)` 한 쌍을 제공합니다. `for index, amount in ...`은 그 두 값을 각각의 변수에 나누어 저장합니다. 존재하는 항목만 처리하므로 범위를 직접 계산하지 않아 마지막 위치를 넘지 않습니다.

```python
# enumerate로 위치 번호와 값을 함께 순회합니다.
visited = []
for index, amount in enumerate(order_amounts):
    visited.append(amount)
    print(index, amount)
print(f"처리 건수: {len(visited)}건")
assert visited == order_amounts
```
