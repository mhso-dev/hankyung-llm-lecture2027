# 클래스

서로 다른 주문 객체가 각자의 상태를 따로 유지하게 만들려면 어떻게 해야 할까요?

## 실행 환경 준비

Python 3.12 내장 기능만 사용합니다. 외부 파일과 이전 노트북의 변수는 필요하지 않습니다. 클래스 본문도 함수와 마찬가지로 콜론 다음 줄을 들여씁니다. 타입 힌트는 07절에서 배우므로 이 노트북의 클래스에는 사용하지 않습니다.

```python
# 표준 Python만 사용하므로 추가 패키지를 설치하지 않습니다.
print("Python 클래스 실습을 시작합니다.")
```

## 기본 실행

`class Order:`는 `Order`라는 새 객체 유형을 정의합니다. `Order("민지")`처럼 클래스 이름을 호출하면 새 인스턴스가 만들어지고 `__init__`이 자동으로 실행됩니다. 호출에서 입력한 `"민지"`는 `customer` 매개변수로 들어갑니다. `self`는 지금 만들고 있는 인스턴스를 가리키므로 `self.customer = customer`는 입력값을 그 객체의 속성에 저장합니다. `self.items = []`는 객체마다 새 빈 목록을 만듭니다.

```python
# 주문 객체를 만들 때마다 고객 이름과 새 상품 목록을 저장합니다.
class Order:
    def __init__(self, customer):
        self.customer = customer
        self.items = []

    def add(self, item):
        self.items.append(item)

first = Order("민지")
first.add("노트")
print(first.customer, first.items)
```

## 중간 결과 관찰

같은 클래스에서 두 인스턴스를 만듭니다. `second.add("연필")`을 호출하면 점 왼쪽의 `second`가 메서드 안의 `self`로 전달되고 문자열은 `item`으로 전달됩니다. `is`는 두 값의 내용이 같은지가 아니라 두 변수가 메모리의 같은 객체를 가리키는지 검사합니다. 두 주문의 목록이 분리되었다면 `first.items is second.items`는 `False`입니다.

`is not`은 두 이름이 서로 다른 객체를 가리키는지 확인합니다. `first.items is not second.items`가 참이면 두 주문의 목록은 별개 객체입니다.

```python
# 두 고객의 상품 목록이 서로 다른 객체인지 확인합니다.
second = Order("준호")
second.add("연필")
print(first.items, second.items)
print("목록을 공유하는가:", first.items is second.items)
assert first.items is not second.items
```

## 값 하나를 바꿔 비교

입력으로 `"지우개"`를 첫 번째 주문의 `add()`에만 전달합니다. 메서드는 `first.items`를 변경하며 별도 값을 반환하지 않습니다. 두 번째 주문의 목록을 출력해 `['연필']`로 유지되는지 확인합니다.

```python
# 민지 주문만 바꾸고 준호 주문은 그대로인지 확인합니다.
first.add("지우개")
print("민지 주문:", first.items)
print("준호 주문:", second.items)
assert second.items == ["연필"]
```

## 실패 재현

클래스 본문의 `items = []`는 인스턴스를 만들 때마다 실행되지 않습니다. 클래스 정의가 실행될 때 목록 하나가 만들어지고, 별도의 인스턴스 속성이 없는 모든 주문이 그 클래스 속성을 찾습니다. 첫 주문을 통해 목록을 변경한 뒤 두 번째 주문에서 같은 내용이 보이는지 확인합니다.

```python
# 클래스에 둔 목록을 두 주문이 공유하는 문제를 재현합니다.
class SharedOrder:
    items = []

bad_first = SharedOrder()
bad_second = SharedOrder()
bad_first.items.append("노트")
print("두 번째 주문의 잘못된 상태:", bad_second.items)
assert bad_second.items == ["노트"]
```

## 수정

가변 상태를 `__init__` 안의 인스턴스 속성으로 옮깁니다. `IsolatedOrder()`를 호출할 때마다 `self.items = []`가 실행되어 새 목록이 해당 인스턴스에 저장됩니다. 같은 인스턴스 수와 변경 순서를 유지하고 목록 생성 위치만 바꾼 뒤 두 번째 주문이 비어 있는지 확인합니다.

```python
# 초기화 메서드에서 각 주문에 별도의 목록을 만듭니다.
class IsolatedOrder:
    def __init__(self):
        self.items = []

good_first = IsolatedOrder()
good_second = IsolatedOrder()
good_first.items.append("노트")
print("두 번째 주문의 수정된 상태:", good_second.items)
assert good_second.items == []
```
