# 함수

입력값을 받아 결제 금액을 계산하고 반환하는 함수를 작성합니다. 함수 정의, 호출, 입력, 처리, 반환을 한 단계씩 확인한 뒤 변수의 범위와 기본값의 주의점을 살펴봅니다.

## 실행 환경 준비

이 노트북은 Python 3.12 표준 기능만 사용합니다. 타입 힌트는 07절에서 배우므로 여기서는 함수의 기본 구조에 집중해 이름과 괄호만 사용합니다. `def`로 함수를 정의하고, 함수 이름 뒤에 소괄호를 붙여 호출합니다.

```python
# amount와 discount_rate를 입력받아 결제 금액을 반환합니다.
def calculate_payment(amount, discount_rate):
    discount = amount * discount_rate
    return round(amount - discount)

print(calculate_payment(78000, 0.1))
```

## 기본 실행

첫 줄 `def calculate_payment(amount, discount_rate):`는 함수를 정의합니다. `amount`와 `discount_rate`는 함수가 받을 입력의 자리 이름인 매개변수입니다. 들여쓴 두 줄은 입력으로 할인액을 계산하는 처리입니다. `return`은 계산 결과를 호출한 곳에 반환합니다. `calculate_payment(amount=78000, discount_rate=0.1)`에서 등호로 이름을 지정한 실제 입력값은 인자이며, 반환된 정수 70200이 `payment`에 저장됩니다. `// 1000`은 나눗셈 결과보다 크지 않은 정수로 내리는 몫 연산입니다. 이 예제의 양수 70200을 1000으로 나눈 70.2에서는 70이 됩니다.

```python
# 반환값을 변수에 저장하면 뒤 계산에서 다시 사용할 수 있습니다.
payment = calculate_payment(amount=78000, discount_rate=0.1)
reward_points = payment // 1000
print(f"결제 금액: {payment}원")
print(f"적립 예정: {reward_points}점")
```

## 중간 결과 관찰

함수 안에서 만든 지역 변수는 함수 밖의 같은 이름을 가진 변수와 별개입니다. 바깥 `label`에는 `"주문"`이 저장되어 있습니다. `make_label()`을 호출하면 안쪽에 별도의 `label`이 만들어지고, 그 값만 반환합니다. 함수가 끝난 뒤 바깥 변수를 출력하면 여전히 `"주문"`입니다.

```python
# 함수 안과 밖의 label은 이름만 같고 범위가 다릅니다.
label = "주문"

def make_label(order_id):
    label = f"주문번호 {order_id}"
    return label

print(make_label("A102"))
print(label)
```

## 값 하나를 바꿔 비교

주문 금액은 78000원으로 고정하고 두 번째 인자만 0.1에서 0.2로 바꿉니다. 같은 정의를 다시 작성하지 않아도 함수 호출마다 다른 입력을 처리해 각각의 결제 금액을 반환합니다.

```python
# 할인율 하나만 바꿔 반환값을 비교합니다.
payment_10 = calculate_payment(78000, 0.1)
payment_20 = calculate_payment(78000, 0.2)
print(f"10% 할인: {payment_10}원")
print(f"20% 할인: {payment_20}원")
```

## 실패 재현

`tags=[]`의 등호 오른쪽은 인자를 생략했을 때 사용할 기본값입니다. 함수 정의가 실행될 때 빈 목록이 한 번 만들어지며, 호출할 때마다 자동으로 새 목록이 생기지 않습니다. 첫 호출이 이 목록에 `신규`를 추가하고 두 번째 호출이 같은 목록에 `배송`을 추가하므로 두 호출의 결과가 섞입니다. 타입 힌트는 아직 사용하지 않습니다.

```python
# 기본 목록이 호출 사이에 공유되는 결과를 관찰합니다.
def add_tag_before(tag, tags=[]):
    tags.append(tag)
    return tags

first_tags = add_tag_before("신규")
first_snapshot = first_tags.copy()
second_tags = add_tag_before("배송")
print(f"첫 호출 시점: {first_snapshot}")
print(f"두 번째 호출 뒤 첫 변수: {first_tags}")
print(f"두 번째 호출 결과: {second_tags}")
print(f"같은 객체: {first_tags is second_tags}")
```

## 수정

기본값을 `None`으로 바꿉니다. `None`은 값이 아직 없음을 나타내는 특별한 값입니다. 인자가 생략되어 `tags is None`이 참일 때 함수 본문에서 빈 목록을 새로 만들기 때문에 각 호출은 서로 다른 목록을 처리하고 반환합니다.

```python
# 호출마다 새 목록을 만들어 이전 호출의 변경을 분리합니다.
def add_tag(tag, tags=None):
    if tags is None:
        tags = []
    tags.append(tag)
    return tags

first_tags = add_tag("신규")
second_tags = add_tag("배송")
print(f"첫 호출: {first_tags}")
print(f"두 번째 호출: {second_tags}")
print(f"같은 객체: {first_tags is second_tags}")
assert first_tags == ["신규"]
assert second_tags == ["배송"]
```
