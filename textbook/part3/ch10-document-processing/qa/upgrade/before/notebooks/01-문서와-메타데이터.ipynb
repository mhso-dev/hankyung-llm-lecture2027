<h3>문서와 메타데이터 준비</h3>
<p><strong>개념</strong> 문서 본문과 출처 정보는 함께 이동해야 합니다. 먼저 실행 위치와 실제 문서 경로를 확인한 뒤 파일명, 확장자, 쪽 번호를 메타데이터로 정리합니다.</p>
<h3>경로 정보와 문서 메타데이터를 구분하기</h3><p><code>Path</code>는 파일 위치를 표현합니다. <code>name</code>·<code>stem</code>·<code>suffix</code>는 저장된 속성을 읽고, <code>is_file()</code>처럼 괄호가 있으면 검사를 실행합니다. 이번 노트북은 LangChain <code>Document</code>를 만들지 않고, 다음 절에서 본문과 묶을 출처 사전까지만 준비합니다.</p><p>메타데이터는 모델이 자동으로 읽는 본문이 아닙니다. 코드가 검색 필터와 출처 표시에 사용하므로, 개정판 문서라면 파일명 외에 문서 ID와 시행일도 별도 키로 저장합니다.</p>

```python
# sys는 현재 노트북 커널의 Python 정보를 제공합니다.
# version.split()[0]은 긴 버전 문자열에서 버전 번호만 고릅니다.
import sys
print("Python:", sys.version.split()[0])
```

<h3>노트북 실행 위치 확인</h3>
<p>이 노트북은 내려받은 실습 폴더 안의 어느 위치에서 실행해도 됩니다. <code>Path.cwd()</code>는 현재 실행 폴더를 나타내며, 아래 셀은 상위 폴더를 차례로 확인해 제공된 PDF를 찾습니다.</p>

```python
# Path는 파일과 폴더 경로를 운영체제에 맞게 다루는 객체입니다.
# parents는 현재 폴더의 상위 폴더들을 가까운 순서로 제공합니다.
from pathlib import Path

repo = None
for candidate in (Path.cwd(), *Path.cwd().parents):
    data_dir = candidate / "labs/data/rag-corpus"
    if data_dir.is_dir():
        repo = candidate
        break
if repo is None:
    raise FileNotFoundError("저장소 안에서 노트북을 실행하세요.")
source = repo / "labs/data/rag-corpus/pdf/easy/E02.pdf"
if not source.is_file():
    raise FileNotFoundError(source)
print("현재 폴더 이름:", Path.cwd().name)
print("실습 문서:", source.relative_to(repo))
```

<h3>경로 객체의 값 관찰</h3>
<p><code>name</code>은 파일명 전체, <code>stem</code>은 확장자를 뺀 이름, <code>suffix</code>는 점을 포함한 확장자입니다. 속성은 객체가 가진 값을 읽는 이름입니다.</p>

```python
# source는 실제 파일을 가리키는 Path 객체입니다.
# type()은 객체의 자료형을, 각 속성은 경로에서 분리한 문자열을 반환합니다.
print("객체:", type(source).__name__)
print("name:", source.name)
print("stem:", source.stem)
print("suffix:", source.suffix)
```

<h3>공통 메타데이터 만들기</h3>
<p>딕셔너리는 이름표인 키와 값의 쌍을 저장합니다. 파일을 읽는 도구가 문자열로 제공한 쪽 번호는 정수로, 쉼표가 든 태그 문자열은 문자열 목록으로 바꿉니다.</p>

```python
# raw는 로더가 반환한 원래 값을 나타내는 딕셔너리입니다.
# int()는 쪽 번호를 정수로, join()은 태그 목록을 쉼표로 연결한 문자열로 바꿉니다.
raw = {"page": "2", "tags": ["배송", "환불"]}
metadata = {
    "doc_id": source.stem,
    "source": source.name,
    "file_type": source.suffix.lstrip(".").lower(),
    "parser": "notebook-demo",
    "page": int(raw["page"]),
    "tags": ",".join(raw["tags"]),
}
print(metadata)
```

<h3>값 하나를 바꾸어 비교</h3>
<p>같은 문서에서 선택한 쪽만 2쪽에서 3쪽으로 바꿉니다. <code>{**metadata, "page": 3}</code>은 원본 딕셔너리를 수정하지 않고 새 딕셔너리를 만듭니다.</p>

```python
# **metadata는 기존 키와 값을 새 딕셔너리에 펼칩니다.
# 뒤에 적은 page 값이 기존 page를 덮어쓰므로 두 결과를 비교할 수 있습니다.
page_2 = metadata
page_3 = {**metadata, "page": 3}
print("기본 선택:", page_2["page"])
print("변경 선택:", page_3["page"])
print("원본 유지:", metadata["page"])
```

<h3>실패 확인</h3>
<p><strong>Before 원인</strong> Python 목록의 두 번째 항목은 위치 <code>1</code>로 고릅니다. 이 위치값을 그대로 출처 쪽 번호에 쓰면 실제 2쪽을 1쪽이라고 표시합니다.</p>

```python
# selected_index는 Python이 두 번째 쪽을 가리킬 때 사용하는 0부터 센 위치입니다.
selected_index = 1
wrong_metadata = {**metadata, "page": selected_index}
actual_page = selected_index + 1

print("실제로 선택한 쪽:", actual_page)
print("잘못 기록한 출처 쪽:", wrong_metadata["page"])
try:
    assert wrong_metadata["page"] == actual_page
except AssertionError:
    print("caught: 선택한 쪽과 출처 표시가 다릅니다.")
```

<p><strong>After 수정</strong> 0부터 센 위치에 <code>1</code>을 더해 사람이 보는 쪽 번호로 바꿉니다. 같은 두 번째 쪽이 출처에도 2쪽으로 기록되는지 확인합니다.</p>

```python
# 사람이 보는 쪽 번호는 1부터 시작하므로 위치값에 1을 더합니다.
page = selected_index + 1
fixed_metadata = {**metadata, "page": page}

assert fixed_metadata["page"] == actual_page
print("실제로 선택한 쪽:", actual_page)
print("수정한 출처 쪽:", fixed_metadata["page"])
```
